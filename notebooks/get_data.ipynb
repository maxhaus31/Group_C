# Get Data

```python
import pandas as pd
import requests
```

## 1. Anual Change in forest area

```python
# Fetch the data.
df = pd.read_csv("https://ourworldindata.org/grapher/annual-change-forest-area.csv?v=1&csvType=full&useColumnShortNames=true", storage_options = {'User-Agent': 'Our World In Data data fetch/1.0'})

# Fetch the metadata
metadata = requests.get("https://ourworldindata.org/grapher/annual-change-forest-area.metadata.json?v=1&csvType=full&useColumnShortNames=true").json()

df.to_csv("../downloads/annual-change-forest-area.csv", index=False)
```
